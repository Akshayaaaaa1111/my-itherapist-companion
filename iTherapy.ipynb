```python
def detect_intervention_type(text):

    text_lower = text.lower()

    if "sleep" in text_lower or "bed" in text_lower:
        return "sleep"
    elif "breath" in text_lower:
        return "breathing"
    elif "ground" in text_lower:
        return "grounding"
    elif "reframe" in text_lower or "negative thought" in text_lower:
        return "reframing"
    elif "mood" in text_lower:
        return "mood"
    else:
        return "general"
```

```python
#Convert Counseling Dataset → Unified Docs
from langdetect import detect, LangDetectException
counseling_docs = []
for row in counseling_ds:
    text = f"{row['Context']} {row['Response']}"
    try:
        if detect(text) != "en":
            continue
    except LangDetectException:
        continue

    counseling_docs.append({
        "content": f"User: {row['Context']}\nTherapist: {row['Response']}",
        "metadata": {
            "source": "counseling",
            "dataset": "Amod/mental_health_counseling_conversations",
            "intervention_type": detect_intervention_type(text),
            "language": "en"
        }
    })
```

```python
#Load Synthetic Function-Calling Dataset --> Query analyzer & function calling
function_docs = []

with open("/content/dataset2_1000_personalized_final.jsonl", "r") as f:
    for line in f:
        item = json.loads(line)

        content = f"""
Intent: {item['intent']}
User Input: {item['user_input']}
Expected Action: {item['function_name']}
""".strip()

        function_docs.append({
            "content": content,
            "metadata": {
                "source": "function",
                "intent": item["intent"]
            }
        })
```

```python
def retrieve_context(query, k=2, threshold=0.6):

    results = collection.query(
        query_texts=[query],
        n_results=k,
        where={"source": {"$eq": "counseling"}}
    )

    docs = results["documents"][0]
    distances = results["distances"][0]

    filtered_docs = [
        doc for doc, dist in zip(docs, distances)
        if dist < threshold
    ]

    return filtered_docs
```

```python
# Query Analyzer (Context-Aware)

def query_analyzer_llm(user_query: str) -> str:

    system_prompt = """
You are a mental health safety classifier.

Classify the user query into ONE of the following:

1. crisis → If user expresses suicidal thoughts, self-harm,
   hopelessness, or immediate danger.
2. personalized → If user refers to their own history,
   previous sessions, personal data, sleep, mood scores.
3. general → All other mental health conversations.

Rules:
- If unsure between crisis and general → choose crisis.
- Return ONLY one word: crisis, personalized, or general.
- Do NOT explain.
"""

    prompt = f"{system_prompt}\n\nUser Query:\n{user_query}\n\nClassification:"

    response = model.generate_content(
        prompt,
        generation_config={"temperature": 0.0}
    )

    result = response.text.strip().lower()

    if result not in ["crisis", "personalized", "general"]:
        return "general"

    return result
```

```python
# Crisis Response

def crisis_response():
    return """
I'm really sorry you're feeling this way.

If you are in immediate danger or thinking about harming yourself,
please contact immediate support.

📞 India Suicide Helpline: 9152987821
📞 Emergency: 112

You deserve real support, and help is available.
"""
```

```python
#to retrieve content form user's history dataset
def retrieve_ehr(user_id):

    results = collection.query(
        query_texts=["user history"],
        n_results=3,
        where={
            "$and": [
                {"source": {"$eq": "ehr"}},
                {"user_id": {"$eq": user_id}}
            ]
        }
    )

    return results["documents"][0]
```

```python
#create new user ID and add information in the existing dataset
import uuid
from datetime import datetime

def initialize_user_profile(user_id):

    existing = collection.get(
        where={
            "$and": [
                {"source": {"$eq": "ehr"}},
                {"user_id": {"$eq": user_id}}
            ]
        }
    )

    if len(existing["ids"]) == 0:

        collection.add(
            ids=[str(uuid.uuid4())],
            documents=["User profile created."],
            metadatas=[{
                "source": "ehr",
                "user_id": user_id,
                "type": "profile_init",
                "timestamp": str(datetime.now())
            }]
        )

        print(f"New user profile created for {user_id}")
```

```python
def store_ehr_interaction(user_id, user_message, model_response):

    collection.add(
        ids=[str(uuid.uuid4())],
        documents=[f"User: {user_message}\nAssistant: {model_response}"],
        metadatas=[{
            "source": "ehr",
            "user_id": user_id,
            "type": "interaction",
            "timestamp": str(datetime.now())
        }]
    )
```

```python
from transformers import pipeline

# load emotion model
emotion_classifier = pipeline(
    "text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    top_k=None
)

def detect_emotion(text):

    results = emotion_classifier(text)

    if isinstance(results[0], list):
        results = results[0]

    results = sorted(results, key=lambda x: x["score"], reverse=True)

    emotion = results[0]["label"]
    confidence = results[0]["score"]

    if confidence > 0.75:
        severity = "high"
    elif confidence > 0.5:
        severity = "medium"
    else:
        severity = "low"

    return emotion, severity
```

```python
import uuid
from datetime import datetime

def store_emotion(user_id, message, emotion, severity):

    emotion_collection.add(

        ids=[str(uuid.uuid4())],

        documents=[message],

        metadatas=[{
            "user_id": user_id,
            "emotion": emotion,
            "severity": severity,
            "timestamp": str(datetime.now())
        }]
    )
```

```python
def get_emotion_history(user_id):

    return emotion_collection.get(
        where={"user_id": user_id}
    )


def format_emotion_trend(results):

    if len(results["metadatas"]) == 0:
        return "No emotional history."

    trend = ""

    for m in results["metadatas"][-5:]:

        trend += f"{m['emotion']} ({m['severity']}), "

    return trend
```

```python
def check_escalation(user_id, threshold=3):
    RISK_EMOTIONS = {"sadness", "fear", "anger", "disgust"}

    results = get_emotion_history(user_id)

    if len(results["metadatas"]) < threshold:
        return False

    last_entries = results["metadatas"][-threshold:]

    high_risk_count = sum(
        1 for m in last_entries
        if m["severity"] == "high" and m["emotion"] in RISK_EMOTIONS
    )

    return high_risk_count == threshold
```

```python
def detect_intent_llm(user_query):

    system_prompt = """
You are an intent classifier for a mental health assistant.

Classify the user query into ONE of the following intents:
- sleep
- stress
- anxiety
- reframing
- mood_tracking
- grounding
- general_support

Return ONLY the intent label.
"""

    prompt = f"{system_prompt}\n\nUser Query:\n{user_query}\n\nIntent:"

    response = model.generate_content(
        prompt,
        generation_config={"temperature": 0.0}
    )

    return response.text.strip().lower()
```

```python
def retrieve_function_by_intent(intent):

    results = collection.get(
        where={
            "$and": [
                {"source": {"$eq": "function"}},
                {"intent": {"$eq": intent}}
            ]
        }
    )

    if len(results["documents"]) == 0:
        return None

    return results["documents"][0]
```

```python
def extract_function_name(function_doc):

    for line in function_doc.split("\n"):
        if "Expected Action:" in line:
            return line.replace("Expected Action:", "").strip()

    return None
```

```python
def execute_function(function_name, user_query):

    function_to_intervention = {
        "offer_grounding_exercise": "grounding",
        "breathing_exercise": "breathing",
        "sleep_hygiene_advice": "sleep",
        "suggest_sleep_hygiene": "sleep",
        "cognitive_reframing": "reframing",
        "suggest_reframing": "reframing",
        "mood_tracking_prompt": "mood",
        "ask_clarifying_question": "general"
    }

    intervention_type = function_to_intervention.get(function_name)

    if not intervention_type:
        print("No intervention mapping found.")
        return []

    results = collection.query(
        query_texts=[user_query],
        n_results=3,
        where={
            "$and": [
                {"source": {"$eq": "counseling"}},
                {"intervention_type": {"$eq": intervention_type}}
            ]
        }
    )

    return results["documents"][0] if results["documents"] else []
```

```python
#temporary
# ==========================================
# Main Entry: Iterative + Routing
# ==========================================

def run_itherapist(initial_query, user_id=None, min_turns=3):
    initialize_user_profile(user_id)

    # Emotion Detection
    emotion, severity = detect_emotion(initial_query)
    store_emotion(user_id, initial_query, emotion, severity)

    # CLASSIFY FIRST
    query_type = query_analyzer_llm(initial_query)
    print("Detected Query Type:", query_type)

    # INTENT DETECTION
    intent = detect_intent_llm(initial_query)
    #print("Detected Intent:", intent)

    # FUNCTION SELECTION (metadata-based)
    function_doc = retrieve_function_by_intent(intent)

    function_output_docs = []
    function_name = None

    if function_doc:
        function_name = extract_function_name(function_doc)
        function_output_docs = execute_function(function_name, initial_query)

    #print("---- FUNCTION DEBUG ----")
    #print("Function Doc Retrieved:", function_doc)
    #print("Function Name:", function_name)
    #print("Function Output Docs:", function_output_docs)
    #print("------------------------")

    # NEW — emotion detection and storage
    emotion, severity = detect_emotion(initial_query)
    store_emotion(user_id, initial_query, emotion, severity)
    if query_type.lower() != "crisis":
        print("Detected emotion:", emotion, severity)

    # CRISIS → IMMEDIATE RESPONSE
    if query_type == "crisis":
        print("\nCRISIS DETECTED")
        print(crisis_response())
        return

    # OTHERWISE → ITERATIVE SESSION
    conversation_history = []
    turn_count = 0
    user_input = initial_query

    while turn_count < min_turns:

        conversation_history.append({"role": "user", "content": user_input})

        # Build retrieval query from full history
        context_query = " ".join(
            [h["content"] for h in conversation_history if h["role"] == "user"] )

        # Personalized path
        if query_type == "personalized" and user_id is not None:
            ehr_docs = retrieve_ehr(user_id)
            rag_docs = retrieve_context(context_query, k=3)
            retrieved_docs = ehr_docs + rag_docs
        else:
            general_docs = retrieve_context(initial_query, k=3)

        # Combine function-controlled RAG with general RAG
        retrieved_docs = []

        # Always retrieve general RAG
        general_docs = retrieve_context(initial_query, k=3)

        retrieved_docs = []

        # Add function-controlled docs if they exist
        if function_output_docs:
            retrieved_docs.extend(function_output_docs)

        # Always add general docs
        retrieved_docs.extend(general_docs)

        # Build clarifying prompt (NO RAG HERE)
        history_text = "\n".join(
            [f"{h['role'].capitalize()}: {h['content']}" for h in conversation_history]
        )

        prompt = f"""
You are iTherapist, a supportive and empathetic mental health assistant.
The context below contains examples of supportive language patterns.
They may include non-clinical opinions.
Do NOT provide medical diagnoses or medication advice.

Conversation so far:
{history_text}

IMPORTANT:
- Ask ONE supportive clarifying question.
- Base your question ONLY on what the user explicitly said.
- Do NOT introduce new symptoms or assumptions.
- Do NOT give advice yet.
"""

        response = model.generate_content(prompt).text

        conversation_history.append({"role": "assistant", "content": response})
        turn_count += 1

        print(f"\niTherapist (Clarifying Question {turn_count}):")
        print(response)

        user_input = input("\nUser: ")

        #emotion_detector
        emotion, severity = detect_emotion(user_input)
        store_emotion(user_id, user_input, emotion, severity)

        print("Detected emotion:", emotion, severity)

    # FINAL RESPONSE AFTER MINIMUM TURNS
    conversation_history.append({"role": "user", "content": user_input})

    context_query = " ".join(
        [h["content"] for h in conversation_history if h["role"] == "user"]
    )

    if query_type == "personalized" and user_id is not None:
        ehr_docs = retrieve_ehr(user_id)
        rag_docs = retrieve_context(context_query, k=3)
        retrieved_docs = ehr_docs + rag_docs
    else:
        retrieved_docs = retrieve_context(context_query, k=3)

    history_text = "\n".join(
        [f"{h['role'].capitalize()}: {h['content']}" for h in conversation_history]
    )

    context = "\n\n".join(retrieved_docs)

    # NEW — retrieve emotional history
    emotion_history = get_emotion_history(user_id)
    emotion_trend = format_emotion_trend(emotion_history)

    final_prompt = f"""
You are iTherapist, an emotionally intelligent and empathetic mental health assistant.
The context below contains examples of supportive language patterns.
They may include non-clinical opinions.
Do NOT provide medical diagnoses or medication advice.

Conversation so far:
{history_text}

User emotional history:
{emotion_trend}

Context:
{context}
Instructions:

1. Acknowledge the user's emotions empathetically.
2. Provide 2–3 practical coping strategies or actionable suggestions.
3. Tailor suggestions based on emotional trend and severity.

Provide a balanced response combining empathy and actionable help.
"""

    final_response = model.generate_content(final_prompt).text
    store_ehr_interaction(user_id, initial_query, final_response)

    print("\nFinal iTherapist Response:")
    print(final_response)
```

```python
# ==========================================
# Main Entry: Iterative + Routing (Clean Version)
# ==========================================

def run_itherapist(initial_query, user_id=None, min_turns=3):

    initialize_user_profile(user_id)

    # ---------------------------
    # 1️⃣ Emotion Detection
    # ---------------------------
    emotion, severity = detect_emotion(initial_query)
    store_emotion(user_id, initial_query, emotion, severity)

    # ---------------------------
    # 2️⃣ Query Classification
    # ---------------------------
    query_type = query_analyzer_llm(initial_query)
    print("Detected Query Type:", query_type)

    # ---------------------------
    # 3️⃣ Intent Detection
    # ---------------------------
    intent = detect_intent_llm(initial_query)

    # ---------------------------
    # 4️⃣ Function Selection
    # ---------------------------
    function_doc = retrieve_function_by_intent(intent)

    function_output_docs = []
    function_name = None

    if function_doc:
        function_name = extract_function_name(function_doc)
        function_output_docs = execute_function(function_name, initial_query)

    #if query_type.lower() != "crisis":
        #print("Detected emotion:", emotion, severity)

    # ---------------------------
    # 5️⃣ Crisis Handling
    # ---------------------------
    if query_type == "crisis":
        print("\nCRISIS DETECTED")
        print(crisis_response())
        return

    # ---------------------------
    # 6️⃣ Iterative Clarification Phase (NO RAG)
    # ---------------------------
    conversation_history = []
    turn_count = 0
    user_input = initial_query

    while turn_count < min_turns:

        conversation_history.append({"role": "user", "content": user_input})

        history_text = "\n".join(
            [f"{h['role'].capitalize()}: {h['content']}" for h in conversation_history]
        )

        clarifying_prompt = f"""
You are iTherapist, a supportive and empathetic mental health assistant.

Conversation so far:
{history_text}

IMPORTANT:
- Ask ONE supportive clarifying question.
- Base your question ONLY on what the user explicitly said.
- Do NOT introduce new symptoms or assumptions.
- Do NOT give advice yet.
"""

        response = model.generate_content(clarifying_prompt).text

        conversation_history.append({"role": "assistant", "content": response})
        turn_count += 1

        print(f"\niTherapist (Clarifying Question {turn_count}):")
        print(response)

        user_input = input("\nUser: ")

        # Emotion detection for each user turn
        emotion, severity = detect_emotion(user_input)
        store_emotion(user_id, user_input, emotion, severity)

        print("Detected emotion:", emotion, severity)

    # ---------------------------
    # 7️⃣ Final Response Phase (Function-Controlled RAG)
    # ---------------------------
    conversation_history.append({"role": "user", "content": user_input})

    context_query = " ".join(
        [h["content"] for h in conversation_history if h["role"] == "user"]
    )

    retrieved_docs = []

    # Personalized memory (EHR)
    if query_type == "personalized" and user_id is not None:
        ehr_docs = retrieve_ehr(user_id)
        retrieved_docs.extend(ehr_docs)

    # Function-controlled RAG
    if function_output_docs:
        retrieved_docs.extend(function_output_docs)

    # General fallback RAG
    general_docs = retrieve_context(context_query, k=2)
    retrieved_docs.extend(general_docs)

    # Limit context size (prevents API 500 errors)
    context = "\n\n".join([doc[:800] for doc in retrieved_docs])

    history_text = "\n".join(
        [f"{h['role'].capitalize()}: {h['content']}" for h in conversation_history]
    )

    # Emotional trend memory
    emotion_history = get_emotion_history(user_id)
    emotion_trend = format_emotion_trend(emotion_history)

    final_prompt = f"""
You are iTherapist, an emotionally intelligent and empathetic mental health assistant.
Do NOT provide medical diagnoses or medication advice.

Conversation so far:
{history_text}

User emotional history:
{emotion_trend}

Context:
{context}

Instructions:
1. Acknowledge the user's emotions empathetically.
2. Provide 2–3 practical coping strategies or actionable suggestions.
3. Tailor suggestions based on emotional trend and severity.
4. Integrate guidance aligned with the selected intervention strategy.

Provide a balanced response combining empathy and actionable help.
"""

    final_response = model.generate_content(final_prompt).text

    # Store final interaction
    store_ehr_interaction(user_id, initial_query, final_response)

    print("\nFinal iTherapist Response:")
    print(final_response)
```

```python
from fastapi import FastAPI
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
import uvicorn

app = FastAPI()

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_methods=["*"],
    allow_headers=["*"],
)

class ChatRequest(BaseModel):
    user_id: str
    session_id: str
    message: str
    conversation_history: list
    turn: int = 0

@app.post("/chat")
def chat(req: ChatRequest):
    # Call your existing iTherapist logic here
    result = run_itherapist_step(
        user_id=req.user_id,
        message=req.message,
        history=req.conversation_history,
        turn=req.turn,
    )
    return {
        "reply": result["reply"],
        "emotion": result["emotion"],
        "severity": result["severity"],
        "turn": result["turn"],
        "is_final": result["turn"] >= 4,
        "function_called": result.get("function_called"),
    }

# ✅ This works in Colab because nest_asyncio is applied
uvicorn.run(app, host="0.0.0.0", port=8000)
```
